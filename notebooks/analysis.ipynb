```python
from difflib import get_close_matches

def fuzzy_merge(series):
    unique_vals = series.unique()
    mapping = {}

    for val in unique_vals:
        match = get_close_matches(val, unique_vals, n=1, cutoff=0.9)
        if match:
            mapping[val] = match[0]

    return series.map(mapping)

# use:
data2['Institute'] = fuzzy_merge(data2['Institute'])
```

```python
def extract_program_features(series: pd.Series):
    series = clean_text(series)

    # ---------------- DEGREE DETECTION ----------------
    degree = []

    for text in series:
        if 'integrated' in text and ('btech' in text or 'b tech' in text):
            degree.append('btech+mtech')
        elif 'bachelor of science' in text or text.startswith('bs'):
            degree.append('bs')
        elif 'btech' in text or 'b tech' in text:
            degree.append('btech')
        elif 'be' in text:
            degree.append('be')
        elif 'mtech' in text:
            degree.append('mtech')
        else:
            degree.append('unknown')

    degree = pd.Series(degree)

    # ---------------- DURATION ----------------
    duration = series.str.extract(r'(\d+)\s*year')[0]

    # ---------------- BRANCH CLEANING ----------------
    branch = series.copy()

    # remove unwanted parts
    branch = branch.str.replace(r'\d+\s*year[s]?', '', regex=True)
    branch = branch.str.replace(r'(btech|b tech|be|mtech|bachelor of science|integrated)', '', regex=True)

    # normalize branches
    branch = branch.str.replace('computer science and engineering', 'cse', regex=False)
    branch = branch.str.replace('electronics and communication engineering', 'ece', regex=False)
    branch = branch.str.replace('mechanical engineering', 'me', regex=False)
    branch = branch.str.replace('civil engineering', 'ce', regex=False)
    branch = branch.str.replace('mathematics', 'maths', regex=False)
    branch = branch.str.replace(r'\s+', ' ', regex=True)

    branch = branch.str.strip()

    return branch, degree, duration
```
