```python
import os
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data=pd.DataFrame()
for file in os.listdir("data"):
    if file.endswith(".csv"):
        df=pd.read_csv(os.path.join("data",file))
        print(f"Loaded {file} with shape {df.shape}")
        data=pd.concat([data, df], ignore_index=True)
        print(f"Current combined shape: {data.shape}")

```

```python
data['Gender']=data['Gender'].replace({"Female-only (including Supernumerary)":"Female-only"})
```

```python
gender_counts = data['Gender'].value_counts().reset_index(name='count')
gender_counts = data['Gender'].value_counts().reset_index(name='count')
sns.barplot(data=gender_counts, x='Gender', y='count')
plt.title('Gender Distribution')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```
